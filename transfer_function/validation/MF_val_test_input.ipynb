```python
# --- time definition ---
T = 10.0
dt = 1e-3
time = np.arange(0, T, dt)

'''
# --- external drive ---
driving_input = {
    'excitatory': {
    'FS': np.ones_like(time) * (network_config['rates']['freq_exc_FS']),
    'RS': np.ones_like(time) * (network_config['rates']['freq_exc_RS'])
    },
    'inhibitory': {
    'FS': np.ones_like(time) * (network_config['rates']['freq_inh_FS']),
    'RS': np.ones_like(time) * (network_config['rates']['freq_inh_RS'])
    }
}

rates = simulate_MF_FS_RS(
    time = time,
    neuron_models = ['FS', 'RS'],
    params = {'FS': params_SI_FS, 'RS': params_SI_RS},
    poly_params = {'FS': data_FS[FS_idx]['polynomial_params'],
                   'RS': data_RS[RS_idx]['polynomial_params']},
    alphas = {'FS': data_FS[FS_idx]['alpha'],
             'RS': data_RS[RS_idx]['alpha']},
    network_config = network_config,
    driving_input = driving_input
)
'''
```

```python
external_input_protocol = {
    'exc_input_FS_freq_Hz': 5.3,
    'exc_input_FS_time_s': [1.0, 2.0],

    'exc_input_RS_freq_Hz': 3.3,
    'exc_input_RS_time_s': [3.0, 4.0],

    'inh_input_FS_freq_Hz': 10.0,
    'inh_input_FS_time_s': [5.0, 6.0],

    'inh_input_RS_freq_Hz': 30.0,
    'inh_input_RS_time_s': [7.0, 8.0],
}
```

```python
driving_input = {
    'excitatory': {
        'FS': np.zeros_like(time),
        'RS': np.zeros_like(time)
    },
    'inhibitory': {
        'FS': np.zeros_like(time),
        'RS': np.zeros_like(time)
    }
}

mapping = {
    ('excitatory', 'FS'): ('exc_input_FS_freq_Hz', 'exc_input_FS_time_s'),
    ('excitatory', 'RS'): ('exc_input_RS_freq_Hz', 'exc_input_RS_time_s'),
    ('inhibitory', 'FS'): ('inh_input_FS_freq_Hz', 'inh_input_FS_time_s'),
    ('inhibitory', 'RS'): ('inh_input_RS_freq_Hz', 'inh_input_RS_time_s'),
}

for (population, neuron_type), (freq_key, time_key) in mapping.items():

    freq = external_input_protocol[freq_key]
    t_start, t_end = external_input_protocol[time_key]

    mask = (time >= t_start) & (time <= t_end)

    driving_input[population][neuron_type][mask] = freq
```

```python
rates = simulate_MF_FS_RS(
    time = time,
    neuron_models = ['FS', 'RS'],
    params = {'FS': params_SI_FS, 'RS': params_SI_RS},
    poly_params = {'FS': data_FS[FS_idx]['polynomial_params'],
                   'RS': data_RS[RS_idx]['polynomial_params']},
    alphas = {'FS': data_FS[FS_idx]['alpha'],
             'RS': data_RS[RS_idx]['alpha']},
    network_config = network_config,
    driving_input = driving_input
)
```

### Plots

```python
fig, axes = plt.subplots(2, 1, figsize=(8, 5), sharex=True)

colors = {"FS": "#e74c3c", "RS": "#27ae60"}

for idx, pop in enumerate(["FS", "RS"]):
    ax = axes[idx]
    ax.plot(time, rates[pop], color=colors[pop], label=f'{pop} firing rate')
    ax.set_ylabel("Firing rate (Hz)")
    ax.legend()

axes[1].set_xlabel("Time (s)")
plt.suptitle("MF Population Firing Rates")
```
